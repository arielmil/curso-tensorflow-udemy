```python
import numpy as np
```

Step funcition:

```python
def stepFunction(soma):
    if (soma >= 1):
        return 1
    return 0

teste = stepFunction(30)

teste
```

Sigmoid

```python
def sigmoidFunction(soma):
    return 1/(1 + np.exp(-soma))

teste = sigmoidFunction(30)

teste
```

Hyperbolic Tangent

```python
def tahnFunction(soma):
    return (np.exp(soma) - np.exp(-soma))/(np.exp(soma) + np.exp(-soma))

teste = tahnFunction(-0.358)

teste
```

ReLU

```python
def reluFunction(soma):
    if soma >= 0:
        return soma
    else:
        return 0
    
teste = reluFunction(-0.350)

teste
```

Linear

```python
def linearFunction(soma):
    return soma

teste = linearFunction(-0.358)

teste
```

Softmax

```python
def softmaxFunction(x):
    ex = np.exp(x)
    return ex / ex.sum()

valores = [7, 2, 1.3]

teste = softmaxFunction(valores)

teste
```

Um exercicio basico de calculo de erros e taxa de acertos

```python
classes = [1, 0, 1, 0]
calculado = [0.3, 0.02, 0.890, 0.320]

errors = []
errorsMse = []
errorsRmse = []
for ind, el in enumerate(classes):

    val = np.abs(el - calculado[ind])
    errors.append(val)

    valPow = (np.pow(val, 2))
    errorsMse.append(valPow)


mae = np.sum(errors)/len(errors)
mse = np.sum(errorsMse)/len(errorsMse)
rmse = np.sqrt(mse)
taxa_acerto = 1 - mae

print("mae:", mae)
print("mse:", mse)
print("rmse:", rmse)
print("taxa_acerto:", taxa_acerto)
```
