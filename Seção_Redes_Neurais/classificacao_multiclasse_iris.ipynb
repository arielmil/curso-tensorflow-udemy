Neste notebook, será feita uma rede neural para classificar plantas em suas determinadas classes baseadas no treinamento em cima de um dataset delas

```python
from sklearn import datasets
iris = datasets.load_iris()

iris
```

iris é composto por três campos: target, data, e feature_names.

Data contém entradas do dataset com dados sobre 150 plantas

```python
iris.data
```

Target contém 3 classes de planta, cada uma das 150 plantas em data, está associada a uma das 3 classes em target.

```python
iris.target
```

feature_names contém informações sobre o que é cada um dos 4 campos em cada uma das 150 entradas na database

```python
iris.feature_names
```

Primeiro pega-se os inputs da rede neural:

```python
X = iris.data

X
```

Depois pega-se os outputs da rede neural:

```python
Y = iris.target

Y
```

Logo em seguida, aplica-se um pre processamento em cima de X, para padronizar ele deixando-o numa escala distribuida normalmente entre 0 e 1

```python
from sklearn.preprocessing import StandardScaler

scaler_x = StandardScaler()
X = scaler_x.fit_transform(X)

X
```

A seguir, faz-se One hot encoding para que cada entrada em Y, tenha 3 campos, o mesmo número de neuronios na camada de saída da rede neural (Já que existem 3 classes possíveis)

```python
from sklearn.preprocessing import OneHotEncoder

onehot = OneHotEncoder(sparse_output=False)

#Reshape de Y de (150, ) (1D) para (150, 1) (2D) para o OneHotEncoder que espera uma entrada 2D
y = Y.reshape(-1, 1)
y.shape
```

```python
y = onehot.fit_transform(y)
y
```

Cada classe foi codificada com um código único por classe (três códigos únicos)

A classe 0 foi codificada como [1, 0, 0],

a classe 1 foi codificada como [0, 1, 0]

e a classe 2 foi codificada como [0, 0, 1].

Assim, a ultima camada composta de 3 neuronios pode outputar qual classe cada entrada em X pertence á.

Faz-se agora a divisão dos dados entre treinamento e teste (70% para treinamento, e 30% para teste):

A divisão é feita aleatóriamente

```python
from sklearn.model_selection import train_test_split

X_treinamento, X_teste, y_treinamento, y_teste = train_test_split(X, y, test_size = 0.3)

print(X_treinamento)
print(y_treinamento)
print(X_teste)
print(y_teste)
```

```python
print(X_treinamento.shape)
print(X_teste.shape)
```

Agora começa-se a definir a rede neural:

```python
import tensorflow as tf
import numpy as np
```

A rede terá 4 neuronios de entrada, um para cada campo de X, as entradas da rede.

Terá 3 neuronios de saída, cada um representando a probabilidade de a entrada pertencer a uma das 3 classes.

Terá uma camada oculta, já que uma já é o suficiente para resolver este problema, com o número de neuronios determinado pela formula:

(neuronios_entrada + neuronios_saida)/2, arredondando para cima caso necessário

```python
neuronios_entrada = X.shape[1]
neuronios_saida = y.shape[1]
neuronios_oculta = int(np.ceil((X.shape[1] + y.shape[1])/2))

print(X.shape[1])
print(y.shape[1])

neuronios_oculta
```

Define-se o dicionário com os pesos de cada camada, e o dicionário dos biases de cada camada

OBS: oculta são os pesos que ficam entre a camada de entrada e a oculta, e saida são os pesos que ficam entre a camada oculta e a de saída.

oculta: Uma matriz 4 x 4 (16 pesos) aonde cada linha está associada ao peso de cada um dos 4 neuronios de entrada, para cada um dos 4 neuronios da camada oculta.

saída: Uma matrix 4x3 (12 pesos) aonde cada linha está associada ao peso de cada um dos 4 neuronios da camada oculta, para cada um dos 3 neuronios da camada de saída.

```python
W = {"oculta": tf.Variable(tf.random.normal([neuronios_entrada, neuronios_oculta])),
     "saida": tf.Variable(tf.random.normal([neuronios_oculta, neuronios_saida]))}

W
```

Bias:

oculta: um vetor de 4 posições (ou matriz 4x1),
cada uma das 4 bias é a bias de um dos 4 neuronios da camada oculta

saída: um vetor de 3 posições (ou matriz 3x1),
cada uma das 3 bias é a bias de um dos 3 neuronios da camada de saída

```python
b = {"oculta": tf.Variable(tf.random.normal([neuronios_oculta])),
     "saida": tf.Variable(tf.random.normal([neuronios_saida]))}

b
```

Converte-se o dataset para tensores para poder ser usado pelo tensorflow que espera trabalhar com tensores ou numpy arrays.

```python
X_treinamento = tf.convert_to_tensor(X_treinamento, dtype = tf.float32, name = 'X')
y_treinamento = tf.convert_to_tensor(y_treinamento, dtype = tf.float32, name = 'y')
X_teste = tf.convert_to_tensor(X_teste, dtype = tf.float32, name = 'X')
y_teste = tf.convert_to_tensor(y_teste, dtype = tf.float32, name = 'y')

print(X_treinamento)
print(y_treinamento)
print(X_teste)
print(y_teste)
```

Define-se o modelo da rede neural atravez da função mlp:

OBS: Usa-se a função relu pois não se quer trabalhar com números negativos (Ver por que)

OBS2: Softmax não é aplicado na ultima camada, pois a função erro já aplica ela.

```python
@tf.function
def mlp(X, W, b):
    camada_oculta_entrada = tf.add(tf.matmul(X, W["oculta"]), b["oculta"])
    camada_oculta_ativacao = tf.nn.relu(camada_oculta_entrada)
    camada_saida_entrada = tf.add(tf.matmul(camada_oculta_ativacao, W["saida"]), b["saida"])
    return camada_saida_entrada
```

Exemplo do treinamento de uma época:

Utiliza-se o otimizador Adam, que tende a ser melhor do que a descida do gradiente simples

```python
var_list = [W['oculta'], W['saida'], b['oculta'], b['saida']]
otimizador = tf.keras.optimizers.Adam(learning_rate = 0.0001)
erro_medio = 0.0
with tf.GradientTape() as tape:
    model = mlp(X_treinamento, W, b)
    error = tf.reduce_mean(tf.nn.softmax_cross_entropy_with_logits(labels = y_treinamento, logits = model))
gradientes = tape.gradient(error, var_list)
otimizador.apply_gradients(zip(gradientes, var_list))
erro_medio += error.numpy()

erro_medio
```

Utilizaremos batch size = 8, ou seja, a rede treinara de 8 em 8 registros em X_treinamento

(Passa 8 registros, faz as atualizações dos pesos, passa mais 8, etc...)

São 13 batches de 8 registros cada

```python
batch_size = 8
num_batches = int(len(X_treinamento) / batch_size)

num_batches
```

Agora utiliza-se a função np.array_split para separar os registros do dataset em num_batches arrays (13 arrays), cada um com o mesmo número de registros (8 registros), com exeção do primeiro array que tem 9 registros, pois o numero de registros em X_treinamento e y_treinamento não é divisivel por 8.

```python
X_batches = np.array_split(X_treinamento, num_batches)
y_batches = np.array_split(y_treinamento, num_batches)
```

```python
X_batches
```

```python
y_batches
```

Agora faz-se o treinamento

OBS: zip(A, B) junta cada elemento de A e B em pares para serem iterados.

```python
@tf.function
def train_step(x_batch, y_batch):
    with tf.GradientTape() as tape:
        logits = mlp(x_batch, W, b)
        loss = tf.reduce_mean(
            tf.nn.softmax_cross_entropy_with_logits(labels=y_batch, logits=logits)
        )
    grads = tape.gradient(loss, var_list)
    otimizador.apply_gradients(zip(grads, var_list))
    
    return loss
```

```python
epochs = 3000

var_list = [W['oculta'], W['saida'], b['oculta'], b['saida']]
otimizador = tf.keras.optimizers.Adam(learning_rate = 0.0001)

for epoch in range(epochs):
    erro_medio = 0.0
    for i in range(num_batches):
        X_batch, y_batch = X_batches[i], y_batches[i]
        loss = train_step(X_batch, y_batch)
        erro_medio += loss
    
    erro_medio = erro_medio / num_batches
    if epoch % 500 == 0:
        print('Época: ' + str((epoch + 1)) + ' erro medio: ' + str(float(erro_medio.numpy())) + ' erros: ' + str(float(loss.numpy())))
```

Agora faz-se testes com a rede neural já treinada para que ela possa classificar novos registros que ela nunca viu no treino:

```python
previsoes_teste = mlp(X_teste, W, b)
previsoes_teste
```

Utiliza-se softmax na ultima camada para converte-la em um vetor de probabilidades, cada um dos 3 neuronios terá uma probabilidade associada a entrada ser de uma das 3 classes do dataset.

```python
previsoes_teste = tf.nn.softmax(previsoes_teste)
previsoes_teste
```

Agora normalizamos as saídas probabilisticas para que cada output da rede neural (Um array de 3 probabilidades), seja normalizado para o valor 0 1 2 representando uma das 3 classes possíveis:

A função abaixo funciona pegando o indice da maior probabilidade em cada registro (um valor por coluna pelo segundo parametro), isso é a mesma forma como as classes foram definidas, como números de 0 a 2.

```python
previsoes_teste = tf.argmax(previsoes_teste, axis = 1).numpy()
previsoes_teste
```

Fazemos a mesma coisa com y_teste para fins de comparação:

```python
y_teste2 = tf.argmax(y_teste, axis = 1).numpy()
y_teste2
```

Por fim, medimos a taxa de acerto:

```python
from sklearn.metrics import accuracy_score
taxa_acerto = accuracy_score(y_teste2, previsoes_teste)
taxa_acerto*100

print("A taxa de acerto é de: "+ str(taxa_acerto*100) + "%.")
```
