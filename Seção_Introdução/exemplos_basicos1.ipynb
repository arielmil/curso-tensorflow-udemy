```python
numero = 2

vetor = [2, 3, 5]
```

```python
import numpy as np
vetor = np.asarray(vetor)
```

```python
#Retorna o maior valor em vetor

vetor.max()
```

```python
#Retorna o índice do maior valor em vetor

vetor.argmax()
```

```python
#Retorna o menor valor em vetor

vetor.min()
```

```python
#Retorna a média dos elementos em vetor

vetor.mean()
```

```python
#Retorna a soma dos elementos em vetor

vetor.sum()
```

```python
#Cria um novo array como: vetor2[i] = numero * vetor[i]

vetor2 = numero * vetor

vetor2
```
