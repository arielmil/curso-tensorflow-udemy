```python
import numpy as np
```

```python
#Cria um array com 10 elementos (Floats), de 1 a 10

array = np.linspace(1, 10, 10)
```

```python
array
```

```python
# Reshapa o array para ele ter o formato de matrix 2 x 5

array = array.reshape(2, 5)
array
```

```python
# Pega o elemento array[1][3] (Linha 1, col 3) 
# Syntaxe: array[L, C] = array[L + 1][C + 1]

array[0, 2]
```

```python
# Pega de todas as linhas, os elementos de 0 a 3 (3o Excluido)

array[:,0:3]
```

```python
# Pega todos os elementos da linha 0

array[0,:]
```

```python
#Cria um array com 10 elementos inteiros de 1 a 11 (11 excluído)

x = np.arange(1,11)
x
```

```python
#Cria um novo array aonde cada elemento y[i] = x[i]^2

y = x**2
y
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
plt.plot(x, y, '*', color = 'red')
```
